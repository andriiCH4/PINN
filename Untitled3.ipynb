```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
import time

# ==============================================================================
# 1. НАСТРОЙКИ И ГИПЕРПАРАМЕТРЫ
# ==============================================================================

# Пути к вашим файлам в Google Drive
BC_PATH = "/content/drive/MyDrive/PINN - 3/bc_nodes.csv"
PINN_DATA_PATH = "/content/drive/MyDrive/PINN - 3/pinn_data.csv"

# Гиперпараметры обучения
LEARNING_RATE = 1e-3
EPOCHS = 10000  # Начните с этого значения, может потребоваться больше
BATCH_SIZE = 2048

# Веса для различных компонентов функции потерь
# Их нужно будет подбирать для достижения наилучшего результата
WEIGHT_DATA = 1.0
WEIGHT_BC = 1.0
WEIGHT_PHYSICS = 0.01 # Вес физики часто делают меньше, особенно в начале

# Архитектура нейросети
layers = [3, 128, 128, 128, 128, 2] # 3 входа (x,y,z), 4 скрытых слоя по 128 нейронов, 2 выхода (CO, NOx)

# ==============================================================================
# 2. ЗАГРУЗКА И ПОДГОТОВКА ДАННЫХ
# ==============================================================================
print("Шаг 1: Загрузка и подготовка данных...")

# Проверяем, доступен ли GPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используемое устройство: {device}")

def load_and_prepare_data(bc_path, pinn_data_path):
    # Загружаем данные
    # .rename(columns=lambda x: x.strip()) - убирает лишние пробелы в названиях колонок
    bc_df = pd.read_csv(bc_path).rename(columns=lambda x: x.strip())
    pinn_df = pd.read_csv(pinn_data_path).rename(columns=lambda x: x.strip())

    # Определяем колонки для входов (координаты) и выходов (эмиссии)
    input_cols = ['x-coordinate', 'y-coordinate', 'z-coordinate']
    output_cols = ['co', 'no']

    # Извлекаем данные
    X_pinn = pinn_df[input_cols].values
    y_pinn = pinn_df[output_cols].values
    X_bc = bc_df[input_cols].values
    y_bc = bc_df[output_cols].values

    # --- Нормализация данных ---
    # Это КРАЙНЕ ВАЖНО для стабильного обучения
    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()

    # Обучаем скейлеры на всех данных (и внутренних, и граничных)
    X_all = np.vstack((X_pinn, X_bc))
    y_all = np.vstack((y_pinn, y_bc))
    scaler_X.fit(X_all)
    scaler_y.fit(y_all)

    # Применяем нормализацию
    X_pinn_scaled = scaler_X.transform(X_pinn)
    y_pinn_scaled = scaler_y.transform(y_pinn)
    X_bc_scaled = scaler_X.transform(X_bc)
    y_bc_scaled = scaler_y.transform(y_bc)

    # --- Конвертация в тензоры PyTorch ---
    X_pinn_t = torch.tensor(X_pinn_scaled, dtype=torch.float32).to(device)
    y_pinn_t = torch.tensor(y_pinn_scaled, dtype=torch.float32).to(device)
    X_bc_t = torch.tensor(X_bc_scaled, dtype=torch.float32).to(device)
    y_bc_t = torch.tensor(y_bc_scaled, dtype=torch.float32).to(device)

    # Для физической потери нам нужны точки коллокации (можно использовать те же точки)
    X_physics_t = X_pinn_t.clone().detach().requires_grad_(True)

    return X_pinn_t, y_pinn_t, X_bc_t, y_bc_t, X_physics_t, scaler_X, scaler_y

X_pinn, y_pinn, X_bc, y_bc, X_physics, scaler_X, scaler_y = load_and_prepare_data(BC_PATH, PINN_DATA_PATH)

print(f"Загружено {X_pinn.shape[0]} точек данных и {X_bc.shape[0]} точек граничных условий.")
print("Данные успешно подготовлены.")

# ==============================================================================
# 3. ОПРЕДЕЛЕНИЕ МОДЕЛИ PINN
# ==============================================================================
print("\nШаг 2: Создание архитектуры PINN...")

class PINN(nn.Module):
    def __init__(self, layers):
        super(PINN, self).__init__()

        self.activation = nn.Tanh() # Tanh - хороший выбор для PINN

        # Создаем слои сети динамически
        self.net = nn.Sequential()
        for i in range(len(layers) - 1):
            self.net.add_module(f"linear_{i}", nn.Linear(layers[i], layers[i+1]))
            if i < len(layers) - 2:
                self.net.add_module(f"activation_{i}", self.activation)

    def forward(self, x):
        return self.net(x)

    def physics_loss(self, x):
        # x - это точки коллокации
        # Получаем предсказания сети в этих точках
        y = self.forward(x)
        co, nox = y[:, 0:1], y[:, 1:2]

        # Вычисляем производные от выходов (co, nox) по входам (x, y, z)
        # Это сердце PINN - автоматическое дифференцирование
        d_co_d_x = torch.autograd.grad(co, x, grad_outputs=torch.ones_like(co), create_graph=True)[0]
        d_nox_d_x = torch.autograd.grad(nox, x, grad_outputs=torch.ones_like(nox), create_graph=True)[0]

        # Разделяем производные по каждой координате
        d_co_dx, d_co_dy, d_co_dz = d_co_d_x[:, 0:1], d_co_d_x[:, 1:2], d_co_d_x[:, 2:3]
        d_nox_dx, d_nox_dy, d_nox_dz = d_nox_d_x[:, 0:1], d_nox_d_x[:, 1:2], d_nox_d_x[:, 2:3]

        # ==================== ВАЖНО: Физические уравнения ====================
        # Здесь вы должны записать остаточные члены (residuals) ваших уравнений переноса.
        # Общий вид уравнения переноса компонента 'C' в стационарном режиме:
        # ∇ ⋅ (ρuC) - ∇ ⋅ (ρD∇C) - R = 0
        # где u - скорость, D - коэф. диффузии, R - скорость реакции.

        # Для простоты примера, мы создадим "заглушку" для физического закона.
        # Например, потребуем, чтобы градиент концентрации CO по X был равен нулю.
        # ВАМ НУЖНО ЗАМЕНИТЬ ЭТО НА ВАШИ РЕАЛЬНЫЕ УРАВНЕНИЯ!
        # ======================================================================

        # Пример-заглушка:
        physics_residual_co = d_co_dx
        physics_residual_nox = d_nox_dx

        # Потеря на физике - это среднеквадратичная ошибка остаточных членов
        loss = torch.mean(physics_residual_co**2) + torch.mean(physics_residual_nox**2)
        return loss

model = PINN(layers).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
loss_fn = nn.MSELoss() # Среднеквадратичная ошибка для потерь на данных

print("Модель PINN успешно создана.")

# ==============================================================================
# 4. ЦИКЛ ОБУЧЕНИЯ
# ==============================================================================
print("\nШаг 3: Запуск обучения...")
start_time = time.time()

for epoch in range(EPOCHS + 1):
    model.train()

    # --- Прямой проход (Forward pass) ---
    # Предсказания для данных и граничных условий
    y_pinn_pred = model(X_pinn)
    y_bc_pred = model(X_bc)

    # --- Вычисление потерь ---
    # 1. Потеря на данных (насколько предсказания соответствуют CFD данным)
    loss_data = loss_fn(y_pinn_pred, y_pinn)

    # 2. Потеря на граничных условиях
    loss_bc = loss_fn(y_bc_pred, y_bc)

    # 3. Потеря на физике (насколько предсказания удовлетворяют уравнениям)
    loss_phys = model.physics_loss(X_physics)

    # Общая взвешенная потеря
    total_loss = (WEIGHT_DATA * loss_data) + \
                 (WEIGHT_BC * loss_bc) + \
                 (WEIGHT_PHYSICS * loss_phys)

    # --- Обратное распространение ошибки (Backward pass) ---
    optimizer.zero_grad()
    total_loss.backward()
    optimizer.step()

    # --- Вывод прогресса ---
    if epoch % 1000 == 0:
        print(f"Эпоха: {epoch}/{EPOCHS} | "
              f"Loss: {total_loss.item():.6f} | "
              f"Data Loss: {loss_data.item():.6f} | "
              f"BC Loss: {loss_bc.item():.6f} | "
              f"Physics Loss: {loss_phys.item():.6f}")

end_time = time.time()
print(f"\nОбучение завершено за {(end_time - start_time):.2f} секунд.")

# ==============================================================================
# 5. ПРОВЕРКА И СОХРАНЕНИЕ МОДЕЛИ
# ==============================================================================
print("\nШаг 4: Проверка и сохранение модели...")
model.eval() # Переводим модель в режим оценки

# Возьмем случайную точку из тестового набора для примера
with torch.no_grad():
    idx = np.random.randint(0, X_pinn.shape[0])
    sample_X = X_pinn[idx:idx+1]

    predicted_scaled = model(sample_X)

    # Возвращаемся к исходному масштабу данных
    predicted_original = scaler_y.inverse_transform(predicted_scaled.cpu().numpy())
    true_original = scaler_y.inverse_transform(y_pinn[idx:idx+1].cpu().numpy())

    print("\n--- Пример предсказания ---")
    print(f"Истинное значение (CO, NOx):   ({true_original[0][0]:.6f}, {true_original[0][1]:.6f})")
    print(f"Предсказанное значение (CO, NOx): ({predicted_original[0][0]:.6f}, {predicted_original[0][1]:.6f})")

# Сохранение модели
MODEL_SAVE_PATH = "/content/drive/MyDrive/PINN - 3/pinn_combustion_model.pth"
torch.save(model.state_dict(), MODEL_SAVE_PATH)
print(f"\nМодель сохранена в: {MODEL_SAVE_PATH}")
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
import time

# ==============================================================================
# 1. CONFIGURATION AND HYPERPARAMETERS
# ==============================================================================
# Paths to your files in Google Drive
BC_PATH = "/content/drive/MyDrive/PINN - 3/bc_nodes.csv"
PINN_DATA_PATH = "/content/drive/MyDrive/PINN - 3/pinn_data.csv"

# Training Hyperparameters
LEARNING_RATE = 1e-4  # A smaller learning rate can be more stable with complex physics
EPOCHS = 15000        # More epochs might be needed for the complex loss to converge
BATCH_SIZE = 4096     # A larger batch size can sometimes help stabilize training

# Weights for the different components of the loss function
# Tuning these is key to getting good results.
WEIGHT_DATA = 1.0
WEIGHT_BC = 1.0
WEIGHT_PHYSICS = 0.01 # Start with a small weight for physics and increase if needed

# Neural Network Architecture
# [input_size, hidden_layer_1, ..., output_size]
layers = [3, 128, 128, 128, 128, 2] # 3 inputs (x,y,z), 4 hidden layers, 2 outputs (CO, NOx)

# ==============================================================================
# 2. DATA LOADING AND PREPARATION
# ==============================================================================
print("Step 1: Loading and Preparing Data...")

# Check if a GPU is available and set the device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

def load_and_prepare_data(bc_path, pinn_data_path):
    """
    Loads, cleans, normalizes, and converts data to PyTorch tensors.
    """
    # Load data and strip whitespace from column names
    try:
        bc_df = pd.read_csv(bc_path).rename(columns=lambda x: x.strip())
        pinn_df = pd.read_csv(pinn_data_path).rename(columns=lambda x: x.strip())
    except FileNotFoundError as e:
        print(f"ERROR: File not found. Make sure your Google Drive is mounted.")
        print(f"Details: {e}")
        return (None,) * 9 # Return nines Nones to prevent crash

    # Define the columns for inputs, outputs, and physical properties
    input_cols = ['x-coordinate', 'y-coordinate', 'z-coordinate']
    output_cols = ['co', 'no']
    physics_cols = ['x-velocity', 'y-velocity', 'z-velocity', 'density', 'viscosity-eff']

    # Extract numpy arrays from dataframes
    X_pinn = pinn_df[input_cols].values
    y_pinn = pinn_df[output_cols].values
    P_pinn = pinn_df[physics_cols].values
    X_bc = bc_df[input_cols].values
    y_bc = bc_df[output_cols].values

    # --- Data Normalization (Crucial for Training Stability) ---
    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()
    scaler_P = MinMaxScaler()

    # Fit scalers on the full dataset to ensure consistent scaling
    X_all = np.vstack((X_pinn, X_bc))
    y_all = np.vstack((y_pinn, y_bc))
    scaler_X.fit(X_all)
    scaler_y.fit(y_all)
    scaler_P.fit(P_pinn) # Physics properties only exist for interior points

    # Apply the scaling
    X_pinn_scaled = scaler_X.transform(X_pinn)
    y_pinn_scaled = scaler_y.transform(y_pinn)
    P_pinn_scaled = scaler_P.transform(P_pinn)
    X_bc_scaled = scaler_X.transform(X_bc)
    y_bc_scaled = scaler_y.transform(y_bc)

    # --- Convert to PyTorch Tensors ---
    X_pinn_t = torch.tensor(X_pinn_scaled, dtype=torch.float32).to(device)
    y_pinn_t = torch.tensor(y_pinn_scaled, dtype=torch.float32).to(device)
    P_pinn_t = torch.tensor(P_pinn_scaled, dtype=torch.float32).to(device)
    X_bc_t = torch.tensor(X_bc_scaled, dtype=torch.float32).to(device)
    y_bc_t = torch.tensor(y_bc_scaled, dtype=torch.float32).to(device)

    # Clone coordinate tensor for physics loss calculation, enabling gradient tracking
    X_physics_t = X_pinn_t.clone().detach().requires_grad_(True)

    return X_pinn_t, y_pinn_t, X_bc_t, y_bc_t, X_physics_t, P_pinn_t, scaler_X, scaler_y, scaler_P

# Load the data
(X_pinn, y_pinn, X_bc, y_bc, X_physics, P_physics,
 scaler_X, scaler_y, scaler_P) = load_and_prepare_data(BC_PATH, PINN_DATA_PATH)

if X_pinn is not None:
    print(f"Loaded {X_pinn.shape[0]} data points and {X_bc.shape[0]} boundary condition points.")
    print("Data successfully prepared.")

# ==============================================================================
# 3. PINN MODEL DEFINITION
# ==============================================================================
if X_pinn is not None:
    print("\nStep 2: Creating PINN architecture...")

class PINN(nn.Module):
    def __init__(self, layers):
        super(PINN, self).__init__()
        self.activation = nn.Tanh()
        # Dynamically create the neural network layers
        self.net = nn.Sequential()
        for i in range(len(layers) - 1):
            self.net.add_module(f"linear_{i}", nn.Linear(layers[i], layers[i+1]))
            if i < len(layers) - 2:
                self.net.add_module(f"activation_{i}", self.activation)

    def forward(self, x):
        return self.net(x)

    def compute_physics_loss(self, x_coords, physics_properties):
        """
        Computes the residual of the governing PDEs.
        This is the core of the "Physics-Informed" part.
        """
        # Get model predictions for CO and NOx
        y_pred = self.forward(x_coords)
        co, nox = y_pred[:, 0:1], y_pred[:, 1:2]

        # Unpack the physical properties
        u, v, w = physics_properties[:, 0:1], physics_properties[:, 1:2], physics_properties[:, 2:3]
        rho = physics_properties[:, 3:4]
        mu_eff = physics_properties[:, 4:5]

        # --- Calculate PDE residual for CO ---
        # Estimate turbulent diffusivity D = mu_eff / (rho * Sc_t)
        Sc_t = 0.8  # Turbulent Schmidt number, a common assumption for gases
        D = mu_eff / (rho * Sc_t + 1e-8) # Add small epsilon to avoid division by zero

        # 1. Convection term: ∇ ⋅ (ρuC)
        d_rho_u_co_dx = torch.autograd.grad(rho * u * co, x_coords, torch.ones_like(co), create_graph=True)[0][:, 0:1]
        d_rho_v_co_dy = torch.autograd.grad(rho * v * co, x_coords, torch.ones_like(co), create_graph=True)[0][:, 1:2]
        d_rho_w_co_dz = torch.autograd.grad(rho * w * co, x_coords, torch.ones_like(co), create_graph=True)[0][:, 2:3]
        convection_co = d_rho_u_co_dx + d_rho_v_co_dy + d_rho_w_co_dz

        # 2. Diffusion term: ∇ ⋅ (ρD∇C)
        grad_co = torch.autograd.grad(co, x_coords, torch.ones_like(co), create_graph=True)[0]
        d_co_dx, d_co_dy, d_co_dz = grad_co[:,0:1], grad_co[:,1:2], grad_co[:,2:3]

        d_term_x_dx = torch.autograd.grad(rho * D * d_co_dx, x_coords, torch.ones_like(d_co_dx), create_graph=True)[0][:,0:1]
        d_term_y_dy = torch.autograd.grad(rho * D * d_co_dy, x_coords, torch.ones_like(d_co_dy), create_graph=True)[0][:,1:2]
        d_term_z_dz = torch.autograd.grad(rho * D * d_co_dz, x_coords, torch.ones_like(d_co_dz), create_graph=True)[0][:,2:3]
        diffusion_co = d_term_x_dx + d_term_y_dy + d_term_z_dz

        # 3. Reaction term (placeholder): R
        # This is highly complex. For this example, we assume it's zero.
        # A real implementation would require a chemistry model (e.g., Arrhenius).
        reaction_co = 0.0

        residual_co = convection_co - diffusion_co - reaction_co

        # --- Calculate PDE residual for NOx (using a simple placeholder for this example) ---
        # TO-DO: Implement the full convection-diffusion equation for NOx, similar to CO.
        residual_nox = torch.autograd.grad(nox, x_coords, torch.ones_like(nox), create_graph=True)[0][:,0:1]

        # The physics loss is the mean squared error of the residuals (we want them to be zero)
        loss = torch.mean(residual_co**2) + torch.mean(residual_nox**2)
        return loss

if X_pinn is not None:
    model = PINN(layers).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
    loss_fn = nn.MSELoss() # Mean Squared Error for data-based losses
    print("PINN model created successfully.")

# ==============================================================================
# 4. TRAINING LOOP
# ==============================================================================
if X_pinn is not None:
    print("\nStep 3: Starting Training...")
    start_time = time.time()

    for epoch in range(EPOCHS + 1):
        model.train()

        # Forward pass: get predictions
        y_pinn_pred = model(X_pinn)
        y_bc_pred = model(X_bc)

        # --- Calculate Losses ---
        # 1. Data Loss (how well the model fits the interior points)
        loss_data = loss_fn(y_pinn_pred, y_pinn)

        # 2. Boundary Condition Loss (how well the model fits the boundary points)
        loss_bc = loss_fn(y_bc_pred, y_bc)

        # 3. Physics Loss (how well the predictions obey the PDE)
        loss_phys = model.compute_physics_loss(X_physics, P_physics)

        # Total weighted loss
        total_loss = (WEIGHT_DATA * loss_data) + \
                     (WEIGHT_BC * loss_bc) + \
                     (WEIGHT_PHYSICS * loss_phys)

        # --- Backward pass and optimization ---
        optimizer.zero_grad()
        total_loss.backward()
        optimizer.step()

        # --- Print Progress ---
        if epoch % 1000 == 0:
            print(f"Epoch: {epoch}/{EPOCHS} | "
                  f"Total Loss: {total_loss.item():.6f} | "
                  f"Data Loss: {loss_data.item():.6f} | "
                  f"BC Loss: {loss_bc.item():.6f} | "
                  f"Physics Loss: {loss_phys.item():.6f}")

    end_time = time.time()
    print(f"\nTraining finished in {(end_time - start_time):.2f} seconds.")

# ==============================================================================
# 5. VALIDATION AND SAVING
# ==============================================================================
if X_pinn is not None:
    print("\nStep 4: Validating and Saving Model...")
    model.eval()

    # --- Example Prediction ---
    with torch.no_grad():
        # Pick a random point from the dataset for a quick test
        idx = np.random.randint(0, X_pinn.shape[0])
        sample_X_scaled = X_pinn[idx:idx+1]
        sample_y_scaled = y_pinn[idx:idx+1]

        predicted_y_scaled = model(sample_X_scaled)

        # Invert scaling to see the results in original units
        predicted_y_original = scaler_y.inverse_transform(predicted_y_scaled.cpu().numpy())
        true_y_original = scaler_y.inverse_transform(sample_y_scaled.cpu().numpy())

        print("\n--- Random Sample Prediction ---")
        print(f"True Value      (CO, NOx): ({true_y_original[0][0]:.6e}, {true_y_original[0][1]:.6e})")
        print(f"Predicted Value (CO, NOx): ({predicted_y_original[0][0]:.6e}, {predicted_y_original[0][1]:.6e})")

    # --- Save the trained model ---
    MODEL_SAVE_PATH = "/content/drive/MyDrive/PINN - 3/pinn_combustion_model_final.pth"
    torch.save(model.state_dict(), MODEL_SAVE_PATH)
    print(f"\nModel state dictionary saved to: {MODEL_SAVE_PATH}")
```
